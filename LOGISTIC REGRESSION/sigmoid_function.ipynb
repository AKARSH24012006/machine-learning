```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

# 1. Generate a simple dataset
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    class_sep=10
)

# 2. Visualize the dataset
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=80)
plt.title("Dataset")
plt.show()

# 3. Define sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 4. Add bias term to X
X_bias = np.insert(X, 0, 1, axis=1)

# 5. Initialize weights (bias, w1, w2)
weights = np.array([1.0, 1.0, 1.0])

# 6. Predict probability using sigmoid
def predict_prob(X_bias, weights):
    z = np.dot(X_bias, weights)
    return sigmoid(z)

# 7. Predict class (0 or 1) from probability
def predict(X_bias, weights):
    probs = predict_prob(X_bias, weights)
    return np.array([1 if p >= 0.5 else 0 for p in probs])

# 8. Extract weights for line equation
bias = weights[0]
w1 = weights[1]
w2 = weights[2]

# 9. Decision boundary line: w1*x1 + w2*x2 + bias = 0
m = -w1 / w2
b = -bias / w2

x_line = np.linspace(min(X[:, 0]) - 1, max(X[:, 0]) + 1, 100)
y_line = m * x_line + b

# 10. Plot data + decision boundary
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='winter', s=80)
plt.plot(x_line, y_line, linewidth=3, label="Sigmoid Linear Boundary (Untrained)")
plt.title("Sigmoid Decision Boundary (Untrained Weights)")
plt.legend()
plt.show()
```
