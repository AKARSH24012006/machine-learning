```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
python
Copy code
# Generate a simple dataset
X, y = make_classification(
    n_samples=100,
    n_features=2,
    n_informative=1,
    n_redundant=0,
    n_classes=2,
    n_clusters_per_class=1,
    random_state=41,
    class_sep=10
)

# Visualize dataset
plt.figure(figsize=(8,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=80)
plt.title("Dataset")
plt.show()
Step Function (Manual)
python
Copy code
def step(z):
    return 1 if z > 0 else 0

def perceptron(X, y, lr=0.1, epochs=100):

    # Insert bias = 1 at column index 0
    X_bias = np.insert(X, 0, 1, axis=1)    # (100, 3)

    # Initialize weights manually = [bias, w1, w2]
    weights = np.ones(X_bias.shape[1])

    # Training loop
    for epoch in range(epochs):
        for i in range(len(y)):
            
            # Linear combination
            z = np.dot(X_bias[i], weights)
            
            # Predicted output (0 or 1)
            y_hat = step(z)
            
            # Calculate error
            error = y[i] - y_hat

            # Perceptron update rule
            weights = weights + lr * error * X_bias[i]

    return weights

weights = perceptron(X, y, lr=0.1, epochs=50)
print("Final Weights:", weights)

bias = weights[0]
w1 = weights[1]
w2 = weights[2]

# Line formula: w1*x1 + w2*x2 + bias = 0
m = -w1 / w2
b = -bias / w2

x_line = np.linspace(min(X[:,0])-1, max(X[:,0])+1, 100)
y_line = m * x_line + b

plt.figure(figsize=(8,6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=80)
plt.plot(x_line, y_line, color='red', linewidth=3, label="Perceptron Line")
plt.title("Perceptron Decision Boundary")
plt.legend()
plt.show()
```
