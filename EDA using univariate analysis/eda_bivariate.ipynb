```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("tips.csv")
```

```python
sns.scatterplot(
    x = "total_bill",
    y = "tip",
    data = df
)
plt.title("total_bill vs tip")
plt.xlabel("total bill")
plt.ylabel("tip")
plt.show()
```

```python
print(df["total_bill"].corr(df["tip"]))
```

```python
print(df.corr(numeric_only=True))
```
