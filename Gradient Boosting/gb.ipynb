```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# -----------------------------
# Sample Dataset
# -----------------------------
X = np.array([[90,8],
              [100,7],
              [110,6],
              [120,9],
              [80,5]])

y = np.array([3,4,8,6,3])

# -----------------------------
# Step 1 : Initial Prediction
# -----------------------------
initial_prediction = np.mean(y)

pred = np.full(len(y), initial_prediction)

print("Initial Prediction:")
print(pred)

# -----------------------------
# Gradient Boosting Parameters
# -----------------------------
learning_rate = 0.1
n_estimators = 3

trees = []

# -----------------------------
# Training
# -----------------------------
for i in range(n_estimators):

    residual = y - pred

    print(f"\nIteration {i+1}")
    print("Residuals:", residual)

    tree = DecisionTreeRegressor(max_depth=2)
    tree.fit(X, residual)

    correction = tree.predict(X)

    pred = pred + learning_rate * correction

    trees.append(tree)

    print("Updated Prediction:", pred)

# -----------------------------
# Final Prediction
# -----------------------------
print("\nFinal Prediction")
print(pred)
```
