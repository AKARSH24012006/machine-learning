```python
import tensorflow as tf
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Conv2D, Flatten, Dense
from keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0

# Reshape for CNN: (samples, height, width, channels)
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

# CNN with VALID Padding (No Padding)

model_valid = Sequential()

model_valid.add(Conv2D(
    filters=32,
    kernel_size=(3, 3),
    strides=1,
    padding="valid",
    activation="relu",
    input_shape=(28, 28, 1)
))

model_valid.add(Conv2D(
    filters=32,
    kernel_size=(3, 3),
    strides=1,
    padding="valid",
    activation="relu"
))

model_valid.add(Flatten())
model_valid.add(Dense(10, activation="softmax"))

model_valid.summary()

# CNN with SAME Padding (Zero Padding)

model_same = Sequential()

model_same.add(Conv2D(
    filters=32,
    kernel_size=(3, 3),
    strides=1,
    padding="same",
    activation="relu",
    input_shape=(28, 28, 1)
))

model_same.add(Conv2D(
    filters=32,
    kernel_size=(3, 3),
    strides=1,
    padding="same",
    activation="relu"
))

model_same.add(Flatten())
model_same.add(Dense(10, activation="softmax"))

model_same.summary()


model_stride = Sequential()

model_stride.add(Conv2D(
    filters=32,
    kernel_size=(3, 3),
    strides=2,
    padding="valid",
    activation="relu",
    input_shape=(28, 28, 1)
))

model_stride.add(Flatten())
model_stride.add(Dense(10, activation="softmax"))

model_stride.summary()
```
