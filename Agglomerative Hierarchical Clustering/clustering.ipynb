```python
from sklearn.datasets import load_iris
# ml library in python having datasets,clustering,regression etc
```

```python
iris = load_iris()
# storing it in a variable named
# now iris have the dataset
```

```python
X = iris.data
# The .data means
# Give me only the input features.

# So
# X
# becomes
# [[5.1 3.5 1.4 0.2]
#  [4.9 3.0 1.4 0.2]
#  ...
# ]
# Shape
# 150 × 4

# Meaning

# 150 flowers
# 4 features each
```

```python
import matplotlib.pyplot as plt
# matplotlib Library for graphs.
# pyplot Module used for plotting.
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
# scipy
# Another Python library.
# Mostly used for

# Mathematics
# Statistics
# Optimization
# Clustering
# so inside cluser it has heirachiacl cluster tools

# linkage

# This creates the hierarchy.

# Without it,

# there is no clustering tree.

# dendrogram

# This simply draws the tree.
```

```python
linked = linkage(X, method="ward")
```

```python
# X


# contains

# 150 flowers

# Each flower starts as its own cluster.

# Initially

# 150 clusters
# method="ward"

# Ward linkage.

# Meaning

# At every step,

# merge the clusters that increase the within-cluster variance the least.

# It tries to create compact clusters.

# linkage()

# Internally does something like

# Cluster1

# Cluster2

# Cluster3

# ...

# ↓

# Calculate distances

# ↓

# Find nearest clusters

# ↓

# Merge

# ↓

# Repeat

# ↓

# One big tree
# What gets stored?

# Everything about every merge.

# linked

# contains

# Who merged

# At what distance

# How many samples

# It's a matrix with shape:

# (n_samples - 1, 4)

# For the Iris dataset:

# 149 × 4

# because 150 samples require 149 merges to end up as one cluster.
```

```python
print(linked)
```

```python
# Example:

# [101 142 0.15 2]

# means:

# Cluster 101 and cluster 142 merged.
# They were 0.15 units apart.
# The new cluster contains 2 samples.
```

```python
plt.figure(figsize=(12,6))
# Creates an empty figure.

# Think of it as opening a blank canvas.

# figsize
# (width,height)

# Measured in inches.

# 12

# Width

# 6

# Height
```

```python
plt.figure(figsize=(12, 6))

dendrogram(linked)

plt.title("Dendrogram")
plt.xlabel("Samples")
plt.ylabel("Euclidean Distance")

plt.show()
# plt.show, It displays the figure.
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)


# AgglomerativeClustering()

# Creates the clustering model.

# Nothing is trained yet.

# n_clusters=3

# Stop merging when only 3 clusters remain.

# If you set:

# n_clusters=2

# You'll get 2 clusters.

# If:

# n_clusters=5

# You'll get 5 clusters.

# linkage="ward"

# Use Ward linkage.

# This is generally the default and most commonly used choice.
```

```python
labels = model.fit_predict(X)

# This line does two things.

# fit()

# Learns the hierarchical clustering from the data.

# predict()

# Assigns each sample to a cluster.

# fit_predict()

# Combines both into one call.

# Equivalent to:

# model.fit(X)
# labels = model.labels_
# labels

# This is a NumPy array.

# Example:

# [1 1 1 1 0 0 2 2 ...]

# Each number is the cluster assigned to that sample.
```

```python
print(labels)
```

```python
plt.scatter(
    X[:,0],
    X[:,1],
    c=labels,
    cmap="viridis"
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Agglomerative Clustering")
plt.show()
```
