```python
import pandas as pd
```

```python
df = pd.read_csv("placement.csv")
```

```python
X = df.iloc[:,0].values.reshape(-1,1)

# X.reshape(-1,1)
# converts
# [7.5,8.2,9.0]
# into:

# [[7.5],
#  [8.2],
#  [9.0]]

y = df.iloc[:,1].values
```

```python
# the x and y are like input and output cgpa[7.3,6.4] and output package[7lpa,8lpa]
# inputs are called as features
# output are called as labels/targets
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=2
)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
#create model
# At this moment:

# No training happened.

# The model knows nothing.

# Think:

# Newborn baby

# No knowledge yet.
```

```python
# Train Model
lr.fit(X_train,y_train) #real emaning of line is : Find the best possible line through the data.
 #fit means learn from data

# Before:

# CGPA → Package relation unknown

# After fit():

# CGPA → Package relation learned


# Model observes:

# Higher CGPA
# =
# Higher Package

# and tries to find the best line.

# Internally sklearn computes:

# Slope (m)

# and

# Intercept (b)

# for: y = mx+b 
```

```python
# After fit():

# lr

# contains:

# m = 2.5

# b = -10

# stored internally.
```

```python
print("Slope =", lr.coef_)

# coef_ means coefficient

# For Linear Regression:

# coefficient = slope

# Suppose output:

# Slope = [2.6]

# Meaning:

# If CGPA increases by 1

# Package increases by 2.6
```

```python
print("Intercept =", lr.intercept_)
```

```python
print("Prediction for 9.4 CGPA =", lr.predict([[9.4]]))
```

```python
#  Import R² Score
from sklearn.metrics import r2_score

# Metric = Evaluation Method

# We need to know:

# How good is model?
```

```python
y_pred = lr.predict(X_test)
```

```python
print("R2 Score =", r2_score(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_train, y_train)
plt.plot(X_train, lr.predict(X_train), color='red')
plt.show()
```
