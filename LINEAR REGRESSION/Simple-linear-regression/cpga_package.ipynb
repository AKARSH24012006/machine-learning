```python
import numpy as np
import pandas as pd
# What is NumPy?

# NumPy = Numerical Python

# Used for:

# Arrays
# Mathematical operations
# Matrix calculations

# What is Pandas?

# Used for:

# CSV files
# Excel files
# Tables/DataFrames
```

```python

class LR:
    def __init__(self):   # FIXED HERE
        self.m = None
        self.b = None 
        #means no slope and no y intercept

# This function trains the model.

    def fit(self, X_train, y_train):
        X_train = np.array(X_train)
        y_train = np.array(y_train)

#   Before:
# [7.5,8.0,8.5]
#   After:
# array([7.5,8.0,8.5])


# These are for calculating slope.
        num = 0
        den = 0

        for i in range(X_train.shape[0]):
            num += (X_train[i] - X_train.mean()) * (y_train[i] - y_train.mean())
            #Σ(x−x̄)(y−ȳ)
            den += (X_train[i] - X_train.mean()) * (X_train[i] - X_train.mean())
            #Σ(x−x̄)²

        # calculate slope
        self.m = num / den
        #calculate intercept
        self.b = y_train.mean() - (self.m * X_train.mean())
        #formula: b = y- mx̄

        print("Slope m:", self.m)
        print("Intercept b:", self.b)

#Prediction Function

    def predict(self, X_test):  #Used to make predictions.
        X_test = np.array(X_test)
        return self.m * X_test + self.b
    # linear regression eq: y=mx+b

```

```python

# -------------------------------
# Load Data
# -------------------------------

df = pd.read_csv("placement.csv")

# iloc : Integer Location

# Column 0:
# CGPA
# Output:
# [8.1,7.8,9.0,8.5]


X = df.iloc[:, 0].values   # CGPA

# Column 1:
# Package
# Output:
# [9.5,8.0,12.0,10.0]

y = df.iloc[:, 1].values   # Package

```

```python

# -------------------------------
# Train-test split
# -------------------------------

# 80% training
# 20% testing
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2
)
#random_state : rnadom different split otherwise evrytime same split if not used
```

```python


# -------------------------------
# Train the model
# -------------------------------

lr = LR()          # Constructor NOW works
lr.fit(X_train, y_train) 
# Calculates:

# m
# b

# Stores inside object  # This WILL PRINT SLOPE & INTERCEPT
```

```python


# -------------------------------
# Prediction
# -------------------------------

# print(lr.predict(X_test[0]))
print(lr.predict(9.4))

from sklearn.metrics import r2_score

y_pred = lr.predict(X_test)
print("R2 score =", r2_score(y_test, y_pred))

#this was just how everything works from scrath so instead of these we use scikit learn 
#as it is 
#fast,optimised,industry used,tested by thousand of developers
```
