```python
# ==========================================
# K-Means Clustering on Iris Dataset
# ==========================================

# Import dataset
from sklearn.datasets import load_iris

# Import KMeans algorithm
from sklearn.cluster import KMeans

# Import matplotlib for visualization
import matplotlib.pyplot as plt
```

```python

# ------------------------------------------
# Load Dataset
# ------------------------------------------
iris = load_iris()

# Features only
X = iris.data
```

```python

# ------------------------------------------
# Create K-Means Model
# ------------------------------------------
model = KMeans(
    n_clusters=3,
    init="k-means++",
    n_init="auto",
    max_iter=300,
    random_state=42
)

# ------------------------------------------
# Train model and predict clusters
# ------------------------------------------
labels = model.fit_predict(X)

# ------------------------------------------
# Centroids
# ------------------------------------------
centroids = model.cluster_centers_
```

```python

print("Cluster Labels:")
print(labels)

print("\nCentroids:")
print(centroids)

# ------------------------------------------
# Visualization
# ------------------------------------------
plt.figure(figsize=(8,6))

plt.scatter(
    X[:,0],
    X[:,1],
    c=labels,
    cmap="viridis",
    s=60
)
```

```python

# Plot Centroids
plt.scatter(
    centroids[:,0],
    centroids[:,1],
    marker="X",
    s=250,
    # s -> size(large X -> easy to see)
    color="red",
    label="Centroids"
    # This name will appear in the legend.
)


plt.title("K-Means Clustering")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.legend()

# Displays the legend.

plt.show()

# Displays the graph.

# ------------------------------------------
# Inertia (WCSS)
# ------------------------------------------
print("\nInertia (WCSS):", model.inertia_)
```

```python

# What is inertia_?

# It stores the Within-Cluster Sum of Squares (WCSS).

# In simple words:

# Calculate the distance from every point to its assigned centroid.
# Square those distances.
# Add them together.

# That total is the Inertia (WCSS).

# Example:

# Point A → Centroid = 2

# 2² = 4

# Point B → Centroid = 3

# 3² = 9

# Point C → Centroid = 1

# 1² = 1

# WCSS = 4 + 9 + 1 = 14

# A smaller WCSS usually means the points are closer to their centroids, which generally indicates better clustering.
```
